```python
import numpy as np
import matplotlib.pyplot as plt

# 定義誤差範圍
errors = np.linspace(-3, 3, 400)  # 誤差從 -5 到 5

# MSE 損失函數
mse_loss = 0.5 * (errors ** 2)

# Smooth L1 Loss 函數
def smooth_l1_loss(errors):
    loss = np.zeros_like(errors)
    mask = np.abs(errors) < 1
    loss[mask] = 0.5 * (errors[mask] ** 2)
    loss[~mask] = np.abs(errors[~mask]) - 0.5
    return loss

smooth_l1 = smooth_l1_loss(errors)

# 設置字體為 Times New Roman
plt.rcParams["font.family"] = "Times New Roman"

# 繪圖
plt.figure(figsize=(10, 6))
plt.plot(errors, mse_loss, label='MSE Loss', color='blue')
plt.plot(errors, smooth_l1, label='Smooth L1-Loss', color='red')

# 標註 y=±1 的點（其實是 x=±1）
plt.axvline(x=1, color='gray', linestyle='--', label='x=1')
plt.axvline(x=-1, color='gray', linestyle='--', label='x=-1')
plt.text(1.2, plt.ylim()[1]*0.4, 'x = 1', color='black', ha='left', va='center', fontsize=14)
plt.text(-1.2, plt.ylim()[1]*0.4, 'x = -1', color='black', ha='right', va='center', fontsize=14)

# 添加標題和標籤
plt.title('Comparison of MSE Loss and Smooth L1-Loss', fontsize=16)
plt.xlabel('Error (y - Q(s,a,θ))', fontsize=14)
plt.ylabel('Loss Value', fontsize=14)
plt.legend()
plt.grid(False)

# 顯示圖表
plt.show()
```
